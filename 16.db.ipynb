```python
import sqlite3

conn = sqlite3.connect('test.db')
cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS member (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    age INTEGER
)
""")
conn.commit()
```

```python
cursor.execute(
    "INSERT INTO member (name, age) VALUES (?, ?)",
    ("홍길동", 20)
)
conn.commit()
```

```python
cursor.execute("SELECT * FROM member")
rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
import pandas as pd

df = pd.read_sql("SELECT * FROM member", conn)
print(df)
```
