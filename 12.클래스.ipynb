```python
class Student:
    school = '한국고등학교'
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_pass(self):
        return self.score >= 60

    def add_score(self, point):
        self.score += point

    def report(self):
        status = '합격' if self.is_pass() else '불합격'
        return f'{self.name}: {self.score}점 ({status})'
```

```python
# 객체 생성 — 클래스 이름을 함수처럼 호출
s1 = Student('홍길동', 55)

# 속성 접근
print(s1.name)          # 홍길동
print(s1.score)         # 55
print(s1.school)        # 한국고등학교 (클래스 변수)

# 메서드 호출
print(s1.is_pass())     # False
s1.add_score(10)        # 점수 변경
print(s1.score)         # 65
print(s1.report())      # 홍길동: 65점 (합격)
```

```python
students = [
    Student('홍길동', 85),
    Student('이순신', 45),
    Student('세종대왕', 92),
]

for s in students:
    print(s.report())

passed = [s.name for s in students if s.is_pass()]
print(passed)   # ['홍길동', '세종대왕']
```

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    @property
    def grade(self):
        if self.score >= 90:
            return 'A'
        elif self.score >= 80:
            return 'B'
        return 'C'

    @staticmethod
    def is_valid_score(score):
        return 0 <= score <= 100
```

```python
s = Student('홍길동', 85)

print(s.grade)                      # B     ← 괄호 없이! 속성처럼 사용
print(Student.is_valid_score(105))  # False ← 객체 없이 클래스로 직접 호출
```
