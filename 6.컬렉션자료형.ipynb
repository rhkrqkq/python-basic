```python
# 리스트 생성
names = ["홍길동", "이순신", "강감찬"]

# 인덱스로 접근
print(names[0])   # 홍길동
print(names[1])   # 이순신
print(names[-1])  # 강감찬 (마지막 요소)

# 리스트 전체 출력
print(names)
```

```python
names = ["홍길동", "이순신", "강감찬"]

# 값 수정
names[1] = "유관순"
print(names)  # ['홍길동', '유관순', '강감찬']

# 값 추가
names.append("안중근")
print(names)  # ['홍길동', '유관순', '강감찬', '안중근']

# 여러 값 한 번에 추가
names.extend(["세종대왕", "신사임당"])
print(names)
```

```python
numbers = [10, 20, 30, 40, 50]

numbers.remove(30)   # 값으로 삭제
print(numbers)       # [10, 20, 40, 50]

numbers.pop(1)       # 인덱스로 삭제
print(numbers)       # [10, 40, 50]

del numbers[0]       # del 키워드로 삭제
print(numbers)       # [40, 50]
```

```python
scores = [95, 80, 100, 70, 85]

scores.sort()
print("오름차순 정렬:", scores)   # [70, 80, 85, 95, 100]

scores.sort(reverse=True)
print("내림차순 정렬:", scores)   # [100, 95, 85, 80, 70]

# reverse()는 정렬이 아니라 현재 순서를 그대로 뒤집는 것
names = ["홍길동", "이순신", "강감찬"]
names.reverse()
print("리스트 뒤집기:", names)   # ['강감찬', '이순신', '홍길동']
```

```python
# 튜플 생성
person = ("홍길동", 30, "서울")

# 인덱스로 접근
print(person[0])   # 홍길동
print(person[1])   # 30
print(person[-1])  # 서울

# 전체 출력
print(person)
```

```python
a = (5)
b = (5,)

print(type(a))  # <class 'int'>
print(type(b))  # <class 'tuple'>
```

```python
person = ("이순신", 45, "전라남도")

# 튜플의 각 요소를 변수로 풀어내기
name, age, region = person

print(name)    # 이순신
print(age)     # 45
print(region)  # 전라남도
```

```python
# 튜플 → 리스트
tuple_data = ("사과", "바나나", "포도")
list_data = list(tuple_data)
print(list_data)  # ['사과', '바나나', '포도']

# 리스트 → 튜플
new_tuple = tuple(list_data)
print(new_tuple)  # ('사과', '바나나', '포도')
```

```python
person = {
    "이름": "홍길동",
    "나이": 30,
    "직업": "개발자"
}

print(person["이름"])   # 홍길동
print(person["직업"])   # 개발자
```

```python
person = {"이름": "이순신", "나이": 45, "직업": "장군"}

# 값 수정
person["나이"] = 46

# 새로운 항목 추가
person["지역"] = "서울"

# 항목 삭제
del person["직업"]

print(person)
# {'이름': '이순신', '나이': 46, '지역': '서울'}
```

```python
person = {"이름": "세종대왕", "직업": "왕", "업적": "훈민정음 창제"}

# 키만 순회
for key in person:
    print(key, ":", person[key])

# 키와 값을 동시에 순회
for key, value in person.items():
    print(f"{key} → {value}")
```

```python
def introduce(이름, 나이):
    print(f"{이름}, {나이}세")

info = {"이름": "홍길동", "나이": 30}

introduce(**info)                 # 홍길동, 30세
introduce(이름="홍길동", 나이=30)   # 위와 동일
```

```python
fruits = {"사과", "바나나", "포도", "사과"}

print(fruits)        # 중복 제거됨 (출력 순서는 실행할 때마다 다를 수 있음)
print(type(fruits))  # <class 'set'>
```

```python
names = ["홍길동", "이순신", "이순신", "강감찬", "홍길동"]
unique_names = set(names)

print(unique_names)          # 중복 제거된 세트 (순서는 다를 수 있음)
print(list(unique_names))    # 다시 리스트로 변환 가능
```

```python
fruits = {"사과", "포도", "바나나"}

# 요소 추가
fruits.add("수박")
print(fruits)

# 여러 요소 추가
fruits.update(["딸기", "오렌지"])
print(fruits)

# 요소 삭제
fruits.remove("포도")
print(fruits)

# 존재하지 않는 요소 삭제 시 에러 없이 처리
fruits.discard("망고")  # 존재하지 않아도 에러 발생하지 않음
```
