# 기본연산자


```python
a = 10
b = 3

print(a + b)   # 더하기 → 13
print(a - b)   # 빼기 → 7
print(a * b)   # 곱하기 → 30
print(a / b)   # 나누기 → 3.333...
print(a // b)  # 몫 → 3
print(a % b)   # 나머지 → 1
print(a ** b)  # 제곱 → 1000
```

```python
A = True
B = False

print(A and B)   # False
print(A or B)    # True
print(not A)     # False
```

```python
# 문자열 슬라이싱 실습

text = "안녕하세요 저는 홍길동입니다."

# 1. 문자열 일부 추출
print(text[0:5])   # '안녕하세요'
print(text[6:8])   # '저는'
print(text[9:12])  # '홍길동'
print(text[-4:])   # '입니다.'

# 2. 특정 간격으로 슬라이싱
print(text[::2])   # 2칸씩 건너뛰며 추출
print(text[::-1])  # 문자열 뒤집기
```

```python
# 정수형 출력
print("당신의 속도는 %d km입니다." % 30)

speed = 30
print("당신의 속도는 %d km입니다." % speed)
print("당신의 속도는 %s km입니다." % speed)  # 문자열로 자동 변환

# 변수 여러 개
name = "홍길동"
age = 30
print("당신의 이름은 %s이고 나이는 %d입니다." % (name, age))
```

```python
# 인덱스 기반
print("당신의 이름은 {0}이고 나이는 {1}입니다.".format("홍길동", 30))

# 순서 생략
print("당신의 이름은 {}이고 나이는 {}입니다.".format("홍길동", 30))

# 키워드 인자 사용
print("당신의 이름은 {name}이고 나이는 {age}입니다.".format(name="홍길동", age=30))
```

```python
name = "홍길동"
age = 30
speed = 30

print(f"{name}님의 나이는 {age}살이고, 속도는 {speed}km입니다.")
```

```python
pi = 3.1415926535
print("원주율은 %.2f 입니다." % pi)           # 소수 둘째자리까지
print("원주율은 {:.2f} 입니다.".format(pi))    # format 방식
print(f"원주율은 {pi:.2f} 입니다.")           # f-string 방식
```

```python
text = "안녕하세요 저는 홍길동입니다."

print(len(text))                # 문자열 길이
print(text.count("홍"))         # 특정 문자 개수
print(text.find("홍"))          # 위치 찾기 (인덱스 반환)
print(text.replace("홍길동", "이중사"))  # 문자열 치환
```

```python
sentence = "파이썬,자바,자바스크립트"
words = sentence.split(",")       # 리스트로 분리
print(words)

joined = " / ".join(words)        # 리스트를 문자열로 결합
print(joined)
```

```python
data = "   공백 제거 실습   "
print("[" + data.strip() + "]")   # 양쪽 공백 제거
print("[" + data.lstrip() + "]")  # 왼쪽 공백 제거
print("[" + data.rstrip() + "]")  # 오른쪽 공백 제거
```
