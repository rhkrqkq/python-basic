```python
f = open('test.txt', 'w', encoding='utf-8')
txt = '안녕하세요'
f.write(txt)
f.close()
```

```python
f = open('test2.txt', 'w', encoding='utf-8')
f.write('안녕하세요.\n')
f.write('반갑습니다.\n')
f.write('저는 홍길동입니다.\n')
f.close()
```

```python
f = open('test2.txt', 'r', encoding='utf-8')
txt = f.readline()
f.close()
print(txt)
```

```python
f = open('test2.txt', 'r', encoding='utf-8')

while True:
    txt = f.readline()
    if txt:
        print(txt, end='')
    else:
        break

f.close()
```
