```python
def introduce(name, country="대한민국"):
    print(f"이름: {name}, 국적: {country}")

introduce("홍길동")
introduce("이순신", "조선")
```

```python
# 일반 함수
def add(a, b):
    return a + b

# 람다 함수
add_lambda = lambda a, b: a + b

print(add_lambda(3, 7))
```

```python
def worker(callback):
    # 공통 절차 …
    result = 10 + 20
    # 필요 시점에 사용자 제공 함수를 호출
    callback(result)

def on_done(value):
    print("결과:", value)

worker(on_done)  # → 결과: 30
```

### 퀴즈) 숫자 맞히기 게임

```python
import random

answer = random.randrange(1, 101)
cnt = 0

while True:
    num = int(input('숫자를 입력하세요'))
    cnt+=1
    if (num < answer):
        print('큰 수를 입력하세요')
    else:
        print('작은수를 입력하세요')

    if num == answer:
        print('정답입니다 총 시도횟수 ' + str(cnt) + '회')
        break

```
