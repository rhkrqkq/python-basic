```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)

for i in range(1, 10):
    if i == 5:
        continue
    print(i)
```

```python
# for문 사용
result = []
for i in range(1, 11) :
  result.append(i)
print(result)
```

```python
# List comprehension 사용
result = [i for i in range(1,11)]
print(result)
```

```python
# 사용자 예외 생성
class UserError(Exception) :
    def __str__(self):
        return "사용자 에러입니다."

def test() :
    print(1)
    raise UserError
    
try :
    test()
except UserError as e:
    print(e)
```
